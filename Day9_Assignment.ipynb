# Day 9 Assignment: Processed E-commerce Dataset

### Creating a combine and completed dataset from orders, customers, and products dataset.

```python
import pandas as pd
import datetime as dt
```

### 1. Loading the Datasets

```python
products_df = pd.read_csv("Day9_Products.csv")
orders_df = pd.read_csv("Day9_Orders.csv")
customers_df = pd.read_csv("Day9_Customers.csv")

# Display the first few rows to verify loading
display(orders_df.head(2))
```

### 2. Demonstrating `concat()`

```python
orders_products = pd.concat([orders_df, products_df], join="outer")
print(orders_products)
```

### 3. Combining DataFrames using `merge()`
Here we have joined Orders and Customers using Customer_ID  and joined that to Products using Product_ID

```python
merged_df = pd.merge(orders_df, customers_df, on='Customer_ID', how='inner')

final_merged_df = pd.merge(merged_df, products_df, on='Product_ID', how='inner')
display(final_merged_df.head(2))
```

### 4. Data Transformation with `apply()`
I have created a new column calle 'Total_Amount' by multiplying quantity with unit price.

```python
final_merged_df['Total_Amount'] = final_merged_df['Quantity'] * final_merged_df['Unit_Price']

final_merged_df['Order_Size_Category'] = final_merged_df['Quantity'].apply(lambda x: 'Bulk' if x >= 5 else 'Standard')
display(final_merged_df[['Quantity', 'Total_Amount', 'Order_Size_Category']].head(3))
```

### 5. DateTime Operations
I have converted the Order_Date column into datetime objects, and extracted day, month, week and date

```python
# Convert to datetime object
final_merged_df['Order_Date'] = pd.to_datetime(final_merged_df['Order_Date'])

# Extract date features
final_merged_df['Order_Month'] = final_merged_df['Order_Date'].dt.month_name()
final_merged_df['Order_Day'] = final_merged_df['Order_Date'].dt.day
final_merged_df['Day_of_Week'] = final_merged_df['Order_Date'].dt.day_name()
display(final_merged_df[['Order_Date', 'Order_Month', 'Day_of_Week']].head(3))
```

### 6. Final Clean-up and Export
Finally, I'll organize the columns into a logical order, removing any redundant data, and export the processed dataset to a CSV file so it's ready to be pushed to GitHub.

```python
columns_order = [
    'Order_ID', 'Order_Date', 'Order_Month', 'Day_of_Week', 
    'Customer_Name', 'City', 'Region', 'Membership_Type',
    'Product_Name', 'Category', 'Brand', 'Unit_Price', 
    'Quantity', 'Total_Amount', 'Order_Size_Category', 
    'Payment_Method', 'Order_Status'
]
processed_df = final_merged_df[columns_order]

processed_df.to_csv('Processed_Ecommerce_Dataset.csv', index=False)
print("Dataset successfully processed and exported to 'Processed_Ecommerce_Dataset.csv'!")
```
